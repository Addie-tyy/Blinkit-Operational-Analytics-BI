```python
import pandas as pd

orders = pd.read_csv('blinkit_data/blinkit_orders.csv')
items = pd.read_csv('blinkit_data/blinkit_order_items.csv')
delivery = pd.read_csv('blinkit_data/blinkit_delivery_performance.csv')
feedback = pd.read_csv('blinkit_data/blinkit_customer_feedback.csv')

print("ORDERS TABLE")
display(orders.head())

print("ORDER ITEMS TABLE")
display(items.head())

print("DELIVERY PERFORMANCE TABLE")
display(delivery.head())

print("CUSTOMER FEEDBACK TABLE")
display(feedback.head())
```

```python
# check for null/missing values in each of these transactions table
# and the total number of data in each of the tables
print("ORDER TABLE")
print(f"count: {len(orders):,}")
print(orders.isnull().sum())

print("ORDER ITEM TABLE")
print(f"count: {len(items):,}")
print(items.isnull().sum())

print("DELIVERY PERFORMANCE TABLE")
print(f"count: {len(delivery):,}")
print(delivery.isnull().sum())

print("CUSTOMER FEEDBACK TABLE")
print(f"count: {len(feedback):,}")
print(feedback.isnull().sum())

```

```python
# master operational CSV
import pandas as pd
from google.colab import files

# rename delivery_time_minutes to SLA_delay_minutes
delivery_clean = delivery[['order_id', 'delivery_partner_id', 'delivery_time_minutes', 'distance_km', 'reasons_if_delayed']].copy()
delivery_clean.rename(columns={'delivery_time_minutes': 'sla_delay_minutes'}, inplace=True)

# left join Delivery, Customer Feedback, and Order Items
df_master = orders.merge(delivery_clean, on='order_id', how='left')
df_master = df_master.merge(feedback[['order_id', 'rating', 'feedback_category', 'sentiment']], on='order_id', how='left')
df_master = df_master.merge(items[['order_id', 'product_id', 'quantity', 'unit_price']], on='order_id', how='left')

df_master.to_csv('blinkit_operational_master.csv', index=False)
print("Consolidation complete! Total rows:", len(df_master))
files.download('blinkit_operational_master.csv')
```
